# Lab 7 - Grafana Dashboard

Goal: build a simple Grafana dashboard from the SQLite database.

## Run Grafana

```bash
docker compose up -d grafana
```

Open: `http://localhost:3123`

SQLite path inside Grafana: `/project-data/transactions.db`

## Example Queries

Total transactions:

```sql
SELECT COUNT(*) AS total_transactions FROM transactions;
```

Risk level breakdown:

```sql
SELECT risk_level, COUNT(*) AS count
FROM predictions
GROUP BY risk_level;
```

High risk percentage:

```sql
SELECT
  CASE
    WHEN COUNT(*) = 0 THEN 0
    ELSE 100.0 * SUM(CASE WHEN risk_level = 'high' THEN 1 ELSE 0 END) / COUNT(*)
  END AS high_risk_percentage
FROM predictions;
```
